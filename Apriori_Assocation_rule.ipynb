```python
import pandas as pd

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
# Step 1: Transaction Dataset
transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Butter'],
    ['Bread', 'Milk', 'Butter'],
    ['Milk', 'Bread'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread', 'Butter']
]

```

```python

# Step 2: Convert transactions into binary format
te = TransactionEncoder()

te_array = te.fit(transactions).transform(transactions)
```

```python
# Step 3: Create DataFrame
df = pd.DataFrame(
    te_array,
    columns=te.columns_
)
print("Transaction Data:")
print(df)
```

```python
# Step 4: Apply Apriori Algorithm
frequent_itemsets = apriori(
    df,
    min_support=0.5,
    use_colnames=True
)

print("\nFrequent Itemsets:")
print(frequent_itemsets)
```

```python

# Step 5: Generate Association Rules
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.6
)
```

```python

# Step 6: Display Important Results
print("\nAssociation Rules:")

print(
    rules[
        [
            'antecedents',
            'consequents',
            'support',
            'confidence',
            'lift'
        ]
    ]
)
```

```python

# Step 7: Sort Rules according to Lift
rules = rules.sort_values(
    by='lift',
    ascending=False
)

print("\nTop Association Rules:")

print(
    rules[
        [
            'antecedents',
            'consequents',
            'support',
            'confidence',
            'lift'
        ]
    ].head(10)
)
```
